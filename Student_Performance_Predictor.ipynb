### Load Sample Student Performance Dataset

We'll create a synthetic dataset to simulate student performance. This dataset will include features like `study_hours`, `attendance_percentage`, `sleep_hours`, and the target variable `marks`.

```python
import pandas as pd
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Number of students
num_students = 100

# Generate synthetic data
study_hours = np.random.normal(loc=10, scale=3, size=num_students) # Average 10 hours, std dev 3
attendance_percentage = np.random.uniform(low=70, high=100, size=num_students) # Between 70% and 100%
sleep_hours = np.random.normal(loc=7, scale=1.5, size=num_students) # Average 7 hours, std dev 1.5

# Introduce some correlation with marks
# Marks will be primarily dependent on study_hours and attendance, with some noise
marks = (study_hours * 3) + (attendance_percentage * 0.5) + (sleep_hours * 1) + np.random.normal(loc=0, scale=5, size=num_students)

# Ensure marks are within a reasonable range (0-100)
marks = np.clip(marks, 0, 100)

# Create DataFrame
df = pd.DataFrame({
    'study_hours': study_hours,
    'attendance_percentage': attendance_percentage,
    'sleep_hours': sleep_hours,
    'marks': marks
})

# Display the first 5 rows of the DataFrame
print("Sample Student Performance Data:")
display(df.head())
```
