### 1. Setup and Authentication
To use the Gemini API, add your API key to the secrets manager in the left panel under the "🔑" icon with the name `GOOGLE_API_KEY`.

```python
!pip install -U -q google-generativeai

import google.generativeai as genai
from google.colab import userdata
import PIL.Image
import requests
from io import BytesIO

# Configure the SDK
try:
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
    genai.configure(api_key=GOOGLE_API_KEY)

    # Using Gemini 2.0 Flash based on the list of available models
    model = genai.GenerativeModel('gemini-2.0-flash')
    print('Model initialized successfully using gemini-2.0-flash.')
except Exception as e:
    print(f'Error: {e}. Please check your API key in the Secrets tab.')
```

### 2. Multi-Modal Execution
We'll fetch a sample image and ask the model a question about it.

```python
# 1. Load a reliable image from Google's documentation
img_url = 'https://storage.googleapis.com/generativeai-downloads/images/scones.jpg'

try:
    response = requests.get(img_url, headers={'User-Agent': 'Mozilla/5.0'})
    response.raise_for_status()
    img = PIL.Image.open(BytesIO(response.content))

    # 2. Define the question
    question = "What is in this image? Describe the food and its arrangement."

    # 3. Generate response
    output = model.generate_content([question, img])

    # 4. Display results
    display(img.resize((400, 300)))
    print(f"\nQuestion: {question}")
    print(f"\nAnswer: {output.text}")
except Exception as e:
    print(f"An error occurred: {e}")
```
