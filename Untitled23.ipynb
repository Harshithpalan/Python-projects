# Product Recommendation Model
This notebook demonstrates building a recommendation system using Collaborative Filtering. We will use the `Surprise` library to train a Singular Value Decomposition (SVD) model on user-item ratings.

```python
!pip install scikit-surprise "numpy<2"
```

```python
import pandas as pd
from surprise import SVD, Dataset, Reader, accuracy
from surprise.model_selection import train_test_split

# 1. Load the MovieLens 100k dataset (built-in to Surprise)
data = Dataset.load_builtin('ml-100k')

# 2. Split into training and testing sets
trainset, testset = train_test_split(data, test_size=0.25)

# 3. Use the SVD algorithm (similar to what Netflix used)
algo = SVD()

# 4. Train the algorithm on the trainset
algo.fit(trainset)

# 5. Make predictions on the testset
predictions = algo.test(testset)

# 6. Compute RMSE (Root Mean Squared Error)
print(f'Model Accuracy (RMSE): {accuracy.rmse(predictions)}')
```
