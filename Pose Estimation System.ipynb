# Pose Estimation System with MediaPipe
This notebook demonstrates how to detect human body joints (hands, legs, head) using the MediaPipe library.

```python
import cv2
import mediapipe as mp
import mediapipe.solutions.pose as mp_pose
import mediapipe.solutions.drawing_utils as mp_drawing
import matplotlib.pyplot as plt
import numpy as np
from urllib.request import urlopen

# Initialize Pose
pose = mp_pose.Pose(static_image_mode=True, min_detection_confidence=0.5)

def estimate_pose(image_source):
    # Handle URL or local path
    if image_source.startswith('http'):
        req = urlopen(image_source)
        arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
        image = cv2.imdecode(arr, -1)
    else:
        image = cv2.imread(image_source)

    if image is None:
        print("Invalid image source.")
        return

    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    results = pose.process(image_rgb)

    if results.pose_landmarks:
        annotated_image = image_rgb.copy()
        mp_drawing.draw_landmarks(
            annotated_image, results.pose_landmarks, mp_pose.POSE_CONNECTIONS)

        plt.figure(figsize=(10, 10))
        plt.imshow(annotated_image)
        plt.axis('off')
        plt.title("Pose Estimation Result")
        plt.show()
    else:
        print("No pose detected.")

# Test with a sample image from the web
sample_url = "https://images.pexels.com/photos/4384679/pexels-photo-4384679.jpeg"
print("Running pose estimation on sample image...")
estimate_pose(sample_url)
```
