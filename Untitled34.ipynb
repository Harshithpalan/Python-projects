## Data Loading

We will use the [MovieLens 100k dataset](https://grouplens.org/datasets/movielens/100k/) for this recommendation system. This dataset contains 100,000 ratings from 943 users on 1682 movies.

```python
import pandas as pd
import requests
import zipfile
import io
import os

# URL for the MovieLens 100k dataset
dataset_url = "http://files.grouplens.org/datasets/movielens/ml-100k.zip"

# Directory to extract the dataset
data_dir = "ml-100k"

# Download the dataset
print(f"Downloading dataset from {dataset_url}...")
response = requests.get(dataset_url)
response.raise_for_status() # Ensure we got a successful response

# Extract the zip file
with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    z.extractall()
print("Dataset downloaded and extracted successfully.")

# Load the u.data file (user-item-rating triplets)
r_cols = ['user_id', 'movie_id', 'rating', 'timestamp']
ratings = pd.read_csv(os.path.join(data_dir, 'u.data'), sep='\t', names=r_cols, encoding='latin-1')

# Load the u.item file (movie titles)
m_cols = ['movie_id', 'title'] + [str(i) for i in range(22)] # There are 22 additional columns for genres etc., we only need movie_id and title for now
movies = pd.read_csv(os.path.join(data_dir, 'u.item'), sep='|', names=m_cols, encoding='latin-1', usecols=[0, 1])

# Load the u.user file (user demographics)
u_cols = ['user_id', 'age', 'gender', 'occupation', 'zip_code']
users = pd.read_csv(os.path.join(data_dir, 'u.user'), sep='|', names=u_cols, encoding='latin-1')

print("DataFrames loaded. Displaying head of each:")
display(ratings.head())
display(movies.head())
display(users.head())
```

## Data Merging and Exploration

To better understand our data and prepare it for building recommendation systems, let's merge the `ratings` and `movies` DataFrames. This will allow us to see which movies were rated and by whom, along with their titles.

```python
# Merge ratings with movie titles
merged_data = pd.merge(ratings, movies, on='movie_id')

print("Merged DataFrame head:")
display(merged_data.head())

print("Merged DataFrame info:")
merged_data.info()

print("Number of unique users:", merged_data['user_id'].nunique())
print("Number of unique movies:", merged_data['movie_id'].nunique())
print("Overall average rating:", merged_data['rating'].mean())
```

## Preparing Data for Collaborative Filtering

For collaborative filtering, we need to transform our merged data into a user-item matrix. In this matrix, rows represent users, columns represent movies, and the values are the ratings given by users to those movies. Missing values (NaN) will indicate movies that a user has not rated.

```python
# Create the user-item matrix using pivot_table
user_movie_matrix = merged_data.pivot_table(index='user_id', columns='title', values='rating')

print("User-Item Matrix head:")
display(user_movie_matrix.head())

print("User-Item Matrix shape:", user_movie_matrix.shape)
print("Sparsity of the matrix (percentage of missing values):")
print(f"{user_movie_matrix.isnull().sum().sum() / (user_movie_matrix.shape[0] * user_movie_matrix.shape[1]):.2%}")
```

## Collaborative Filtering (SVD)

Collaborative filtering is a technique that can filter out items that a user might like on the basis of reactions by similar users. It works by collecting and analyzing a large amount of information on users' behaviors, activities or preferences and predicting what users will like based on their similarity to other users.

We'll use a model-based collaborative filtering approach using Singular Value Decomposition (SVD). SVD is a matrix factorization technique that can reduce the number of features of a data set by reducing the space dimension from N to K where K < N.

For implementing SVD, we will use the `surprise` library, which provides various ready-to-use recommendation algorithms.

```python
import numpy as np
from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split, cross_validate

# We need to explicitly define the Reader object. The `rating_scale` parameter is important.
reader = Reader(rating_scale=(1, 5))

# Load data into the surprise dataset format
# We'll use the 'ratings' DataFrame directly for this.
# The order of columns should be user, item, rating
surprise_data = Dataset.load_from_df(ratings[['user_id', 'movie_id', 'rating']], reader)

# Split data into training and test sets
trainset, testset = train_test_split(surprise_data, test_size=0.25, random_state=42)

# Use the SVD algorithm
algo = SVD(random_state=42)

# Train the algorithm on the trainset
algo.fit(trainset)

# Make predictions on the testset
predictions = algo.test(testset)

# Evaluate the predictions (e.g., using RMSE and MAE)
from surprise import accuracy
accuracy.rmse(predictions)
accuracy.mae(predictions)

print("\nExample predictions for a specific user and movie (if rated):")
# Let's pick a user and movie from the original ratings to see a prediction
# For example, user_id 196 rated movie_id 242 with 3
user_id_example = 196
movie_id_example = 242
actual_rating_example = ratings[(ratings['user_id'] == user_id_example) & (ratings['movie_id'] == movie_id_example)]['rating'].iloc[0]
predicted_rating_example = algo.predict(uid=user_id_example, iid=movie_id_example).est

print(f"User {user_id_example} rated movie {movie_id_example} (Kolya (1996)) with actual rating: {actual_rating_example}")
print(f"Predicted rating for User {user_id_example} on Movie {movie_id_example}: {predicted_rating_example:.2f}")

# Let's predict a rating for a movie that user 1 has NOT rated
# Find a movie user 1 has not rated
user1_rated_movies = ratings[ratings['user_id'] == 1]['movie_id'].unique()
all_movies_ids = ratings['movie_id'].unique()
unrated_movie_for_user1 = np.setdiff1d(all_movies_ids, user1_rated_movies)[0] # Get the first unrated movie

# Get the title for this unrated movie
unrated_movie_title = movies[movies['movie_id'] == unrated_movie_for_user1]['title'].iloc[0]

predicted_unrated_rating = algo.predict(uid=1, iid=unrated_movie_for_user1).est
print(f"\nPredicted rating for User 1 on unrated movie '{unrated_movie_title}' (movie_id {unrated_movie_for_user1}): {predicted_unrated_rating:.2f}")
```
