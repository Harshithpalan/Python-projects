```python
def calculate_house_price(m, x, b):
  """
  Calculates house price using the linear equation y = mx + b.

  Args:
    m (float): The slope (e.g., price per square foot).
    x (float): The independent variable (e.g., square footage).
    b (float): The y-intercept (e.g., base price).

  Returns:
    float: The calculated house price.
  """
  return m * x + b
```

Let's demonstrate how to use this function with an example.

```python
# Example usage:
# Assume a house has 1500 square feet (x)
# The price per square foot is $200 (m)
# There's a base price of $50,000 (b)

m_value = 200  # dollars per square foot
x_value = 1500 # square feet
b_value = 50000 # base price

house_price = calculate_house_price(m_value, x_value, b_value)
print(f"The estimated house price is: ${house_price:,.2f}")
```
