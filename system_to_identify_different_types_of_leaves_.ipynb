# Leaf Identification Deep Learning System
This notebook implements a CNN to classify different types of leaves using Transfer Learning with MobileNetV2.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np
import os

# Configuration
BATCH_SIZE = 32
IMG_SIZE = (224, 224)
EPOCHS = 10
```

### 1. Data Preparation
We will use a placeholder for the dataset path. You can upload your dataset to Colab or link it from Drive.

```python
# Example: Loading data from a directory
# dataset_path = 'path_to_your_leaf_images'

# For demonstration, we create a dummy dataset or use tf.keras.utils.image_dataset_from_directory
# if the path exists.

def load_data(data_dir):
    train_ds = tf.keras.utils.image_dataset_from_directory(
        data_dir,
        validation_split=0.2,
        subset='training',
        seed=123,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE)

    val_ds = tf.keras.utils.image_dataset_from_directory(
        data_dir,
        validation_split=0.2,
        subset='validation',
        seed=123,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE)

    return train_ds, val_ds

print("Data loading functions ready. Please provide a directory path to execute classification.")
```

### 2. Model Architecture
We use Transfer Learning with **MobileNetV2** for high efficiency and accuracy on mobile-like tasks.

```python
def build_model(num_classes):
    base_model = tf.keras.applications.MobileNetV2(input_shape=IMG_SIZE + (3,),
                                               include_top=False,
                                               weights='imagenet')
    base_model.trainable = False # Freeze base layers

    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dropout(0.2),
        layers.Dense(num_classes, activation='softmax')
    ])

    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

model = build_model(num_classes=10) # Adjust based on your dataset
model.summary()
```

```python
### 3. Training and Evaluation

# Placeholder for training.
# To run this, you would need to execute:
# train_ds, val_ds = load_data('your_path_here')
# history = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS)

def plot_history(history):
    acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']
    loss = history.history['loss']
    val_loss = history.history['val_loss']
    epochs_range = range(len(acc))

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(epochs_range, acc, label='Training Accuracy')
    plt.plot(epochs_range, val_acc, label='Validation Accuracy')
    plt.legend(loc='lower right')
    plt.title('Training and Validation Accuracy')

    plt.subplot(1, 2, 2)
    plt.plot(epochs_range, loss, label='Training Loss')
    plt.plot(epochs_range, val_loss, label='Validation Loss')
    plt.legend(loc='upper right')
    plt.title('Training and Validation Loss')
    plt.show()

print("Training and visualization logic is ready.")
```
