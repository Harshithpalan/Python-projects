```python
import pandas as pd

# Assuming the data file is named 'smart_water_meter_data.csv'
# If the file name is different, please modify this line.
# For example: df = pd.read_csv('/content/your_data_file.csv')
try:
    df = pd.read_csv('/tmp/smart_water_meter_data.csv')
except FileNotFoundError:
    print("Error: 'smart_water_meter_data.csv' not found. Please upload the data file or provide the correct path/filename.")
    # As a fallback for demonstration, creating a dummy dataframe if file not found
    data = {
        'timestamp': pd.to_datetime(pd.date_range(start='2023-01-01', periods=100, freq='H')),
        'meter_id': [f'M{i}' for i in range(1, 11)] * 10,
        'consumption': [i * 0.5 + 10 + (i % 5) * 2 for i in range(100)]
    }
    df = pd.DataFrame(data)
    print("Created a dummy DataFrame for demonstration.")

df.info()
df.head()

print("\nMissing values per column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

**Reasoning**:
To prepare the data for time-series analysis, I will first set the 'timestamp' column as the DataFrame's index. Then, I will resample the data to a daily frequency, summing the 'consumption' for each day, and display the first few rows of the resampled data.



```python
df_resampled = df.set_index('timestamp').resample('D')['consumption'].sum().to_frame()
print("Resampled data (daily consumption):")
print(df_resampled.head())
```
