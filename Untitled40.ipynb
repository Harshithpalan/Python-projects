### Sample Data Schema for Air Quality Prediction

This schema includes typical features used for predicting Air Quality Index (AQI), combining meteorological data with pollutant concentrations. The `AQI` column is the target variable to be predicted.

**Input Features:**
*   **`timestamp`**: (Datetime) Date and time of the observation.
*   **`latitude`**: (Float) Latitude of the monitoring station.
*   **`longitude`**: (Float) Longitude of the monitoring station.
*   **`temperature_c`**: (Float) Ambient temperature in Celsius.
*   **`humidity_percent`**: (Float) Relative humidity in percentage.
*   **`wind_speed_m_s`**: (Float) Wind speed in meters per second.
*   **`wind_direction_deg`**: (Float) Wind direction in degrees (0-360).
*   **`pressure_hpa`**: (Float) Atmospheric pressure in hectopascals.
*   **`pm2_5_ug_m3`**: (Float) Concentration of Particulate Matter 2.5 in micrograms per cubic meter.
*   **`pm10_ug_m3`**: (Float) Concentration of Particulate Matter 10 in micrograms per cubic meter.
*   **`o3_ppb`**: (Float) Ozone concentration in parts per billion.
*   **`co_ppm`**: (Float) Carbon Monoxide concentration in parts per million.
*   **`so2_ppb`**: (Float) Sulfur Dioxide concentration in parts per billion.
*   **`no2_ppb`**: (Float) Nitrogen Dioxide concentration in parts per billion.

**Target Variable:**
*   **`AQI`**: (Integer) Air Quality Index value. This is what the model will predict.
