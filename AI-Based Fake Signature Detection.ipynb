### 1. Concepts: Siamese Networks
A Siamese Network consists of two identical subnetworks that share the same weights. They take two different images as input and produce feature vectors. The distance between these vectors indicates whether the images are 'similar' (genuine-genuine pair) or 'dissimilar' (genuine-forged pair).
