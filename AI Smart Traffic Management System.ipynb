### Example: Training a Simple ML Model

This example demonstrates a basic machine learning workflow using a synthetic dataset for a classification task. We'll:
1.  Generate synthetic data.
2.  Split the data into training and testing sets.
3.  Train a simple classification model (Logistic Regression).
4.  Make predictions and evaluate the model.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1. Generate synthetic data
X, y = make_classification(n_samples=1000, n_features=20, n_informative=10, n_redundant=5, random_state=42)
df = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(X.shape[1])])
df['target'] = y

print("First 5 rows of the synthetic dataset:")
display(df.head())
print("\nTarget distribution:")
display(df['target'].value_counts())
```

Now, let's split the data into training and testing sets, train a Logistic Regression model, and evaluate its performance.

```python
# 2. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"Training data shape: {X_train.shape}")
print(f"Testing data shape: {X_test.shape}")

# 3. Train a simple classification model (Logistic Regression)
model = LogisticRegression(random_state=42, solver='liblinear')
model.fit(X_train, y_train)

print("\nModel training complete.")

# 4. Make predictions on the test set
y_pred = model.predict(X_test)

# 5. Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"\nModel Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(report)
```
