### Rainfall Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set a random seed for reproducibility
np.random.seed(42)

# Generate sample rainfall data for 30 days (in mm)
dates = pd.to_datetime(pd.date_range(start='2023-01-01', periods=30, freq='D'))
rainfall_mm = np.random.normal(loc=10, scale=5, size=30)
rainfall_mm[rainfall_mm < 0] = 0 # Rainfall cannot be negative

data = {'Date': dates, 'Rainfall_mm': rainfall_mm}
df = pd.DataFrame(data)

print("Sample Rainfall Data:")
display(df.head())
```

#### Basic Statistics

```python
print("Rainfall Statistics:")
display(df['Rainfall_mm'].describe())

# Calculate additional statistics
mean_rainfall = df['Rainfall_mm'].mean()
median_rainfall = df['Rainfall_mm'].median()
max_rainfall = df['Rainfall_mm'].max()
min_rainfall = df['Rainfall_mm'].min()

print(f"\nMean Rainfall: {mean_rainfall:.2f} mm")
print(f"Median Rainfall: {median_rainfall:.2f} mm")
print(f"Maximum Rainfall: {max_rainfall:.2f} mm")
print(f"Minimum Rainfall: {min_rainfall:.2f} mm")
```

#### Visualizing Rainfall Data

```python
plt.figure(figsize=(12, 6))
sns.lineplot(x='Date', y='Rainfall_mm', data=df)
plt.title('Daily Rainfall Over 30 Days')
plt.xlabel('Date')
plt.ylabel('Rainfall (mm)')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.histplot(df['Rainfall_mm'], bins=10, kde=True)
plt.title('Distribution of Rainfall')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
