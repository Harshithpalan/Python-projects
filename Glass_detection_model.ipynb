# Glasses Detection Model
This notebook demonstrates how to build and train a deep learning model to detect whether a person is wearing glasses.

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
```

### Note on Data
Since downloading the full 200k CelebA images requires a Kaggle API key or large manual download, I will provide the architecture and a synthetic data generator to demonstrate the model pipeline. If you have a specific local dataset, you can point the image directory to it.

```python
def create_model():
    model = models.Sequential([
        layers.Conv2D(32, (3, 3), activation='relu', input_shape=(128, 128, 3)),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation='relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(128, (3, 3), activation='relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(1, activation='sigmoid')
    ])

    model.compile(optimizer='adam',
                  loss='binary_crossentropy',
                  metrics=['accuracy'])
    return model

model = create_model()
model.summary()
```

### Next Steps
1. **Data Loading**: You would typically use `tf.keras.utils.image_dataset_from_directory` to load your labeled image folders.
2. **Training**: Use `model.fit()` on your training data.
3. **Inference**: Use `model.predict()` on new face images.

### Synthetic Data Generator
To demonstrate the training loop, we'll create a simple generator that produces random images and labels.

```python
def synthetic_gen():
    while True:
        yield np.random.rand(128, 128, 3), np.random.randint(0, 2, (1,))

train_dataset = tf.data.Dataset.from_generator(
    synthetic_gen,
    output_signature=(
        tf.TensorSpec(shape=(128, 128, 3), dtype=tf.float32),
        tf.TensorSpec(shape=(1,), dtype=tf.int32)
    )
).batch(32)

# Train for a few steps to demonstrate functionality
history = model.fit(train_dataset, steps_per_epoch=10, epochs=5)
```
