## 1. Data Loading

I will use the `datasets` library to load the AG News dataset, which is a common dataset for text classification.

```python
# Install the datasets library if you haven't already
!pip install datasets

from datasets import load_dataset

# Load the AG News dataset
dataset = load_dataset('ag_news')

# Display the dataset structure
print(dataset)
```

The dataset is loaded and split into training and test sets. Let's inspect the first few examples of the training set.

```python
# Display the first few examples of the training set
import pandas as pd

# Convert to pandas DataFrame for easier viewing
train_df = pd.DataFrame(dataset['train'])
test_df = pd.DataFrame(dataset['test'])

print("Training set head:")
display(train_df.head())

print("\nTest set head:")
display(test_df.head())
```

```python
# Display basic information about the dataset
print("Training set info:")
train_df.info()

print("\nTest set info:")
test_df.info()
```

## 2. Data Preprocessing

We will clean the text and use TF-IDF vectorization to convert it into a numerical format suitable for machine learning.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import re

def clean_text(text):
    # Remove non-alphabetic characters and lowercase
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text.lower()

# Apply cleaning to the text data
train_df['text_cleaned'] = train_df['text'].apply(clean_text)
test_df['text_cleaned'] = test_df['text'].apply(clean_text)

# Initialize TF-IDF Vectorizer
tfidf = TfidfVectorizer(max_features=5000, stop_words='english')

# Fit and transform the data
X_train = tfidf.fit_transform(train_df['text_cleaned'])
X_test = tfidf.transform(test_df['text_cleaned'])

y_train = train_df['label']
y_test = test_df['label']

print(f"Training features shape: {X_train.shape}")
```

## 3. Model Training and Evaluation

Now we train a Naive Bayes model and evaluate its accuracy on the test set.

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# Train Multinomial Naive Bayes
model = MultinomialNB()
model.fit(X_train, y_train)

# Predict and Evaluate
y_pred = model.predict(X_test)

print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['World', 'Sports', 'Business', 'Sci/Tech']))
```

## 3. Model Training and Evaluation

We will train a Multinomial Naive Bayes classifier and evaluate it using a classification report and accuracy score.

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# Initialize the model
nb_model = MultinomialNB()

# Train the model
nb_model.fit(X_train, y_train)

# Make predictions
y_pred = nb_model.predict(X_test)

# Evaluate the model
print(f"Accuracy Score: {accuracy_score(y_test, y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['World', 'Sports', 'Business', 'Sci/Tech']))
```
