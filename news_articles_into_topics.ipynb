### 1. Import Libraries and Load Dataset
We will use `scikit-learn` to fetch the news articles and perform the classification.

```python
import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# Define the categories we want to classify
categories = ['rec.sport.baseball', 'sci.electronics', 'comp.graphics', 'talk.politics.mideast']

# Load the dataset
newsgroups = fetch_20newsgroups(subset='all', categories=categories, remove=('headers', 'footers', 'quotes'))

print(f"Total articles loaded: {len(newsgroups.data)}")
```

### 2. Preprocessing and Vectorization
Machine learning models require numerical input. We use `TfidfVectorizer` to convert text into a matrix of TF-IDF features.

```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(newsgroups.data, newsgroups.target, test_size=0.2, random_state=42)

# Initialize TF-IDF Vectorizer
vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)

# Transform the text data
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Vectorization complete.")
```

### 3. Model Training and Evaluation
We will train a Multinomial Naive Bayes model, which is highly effective for text classification.

```python
# Initialize and train the classifier
clf = MultinomialNB(alpha=0.1)
clf.fit(X_train_tfidf, y_train)

# Make predictions
y_pred = clf.predict(X_test_tfidf)

# Evaluation
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}\n")
print(classification_report(y_test, y_pred, target_names=newsgroups.target_names))
```

### 4. Test with Custom News Snippet
Let's see how the model categorizes a new piece of text.

```python
def predict_category(text):
    test_tfidf = vectorizer.transform([text])
    prediction = clf.predict(test_tfidf)
    return newsgroups.target_names[prediction[0]]

sample_news = "The pitcher threw a perfect curveball to strike out the final batter."
print(f"Text: '{sample_news}'")
print(f"Predicted Topic: {predict_category(sample_news)}")
```
