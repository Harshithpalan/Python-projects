### Load Resume Text Files

This function will iterate through a given directory, read all `.txt` files, and return their content as a list of strings.

```python
import os

def load_resume_texts(directory_path):
    """
    Loads the content of all .txt files from a specified directory.

    Args:
        directory_path (str): The path to the directory containing resume text files.

    Returns:
        list: A list of strings, where each string is the content of a resume file.
    """
    resume_texts = []
    if not os.path.isdir(directory_path):
        print(f"Error: Directory not found at {directory_path}")
        return resume_texts

    for filename in os.listdir(directory_path):
        if filename.endswith('.txt'):
            filepath = os.path.join(directory_path, filename)
            try:
                with open(filepath, 'r', encoding='utf-8') as f:
                    resume_texts.append(f.read())
            except Exception as e:
                print(f"Error reading {filepath}: {e}")
    return resume_texts
```

#### Example Usage

First, let's create a dummy directory and some resume text files to demonstrate the function.

```python
# Create a dummy directory and some resume files for demonstration
dummy_dir = 'dummy_resumes'
os.makedirs(dummy_dir, exist_ok=True)

with open(os.path.join(dummy_dir, 'resume_1.txt'), 'w') as f:
    f.write('This is the content of resume 1. It mentions Python and Data Science.')

with open(os.path.join(dummy_dir, 'resume_2.txt'), 'w') as f:
    f.write('Resume 2 talks about Machine Learning and Java programming.')

with open(os.path.join(dummy_dir, 'another_file.pdf'), 'w') as f:
    f.write('This is a dummy PDF content, which should be ignored.') # This won't be read

print(f"Created dummy resume files in '{dummy_dir}'")

# Now, let's use the function to load these resumes
resumes = load_resume_texts(dummy_dir)

print(f"\nLoaded {len(resumes)} resume(s).")
for i, resume_text in enumerate(resumes):
    print(f"Resume {i+1} content: {resume_text[:50]}...") # Displaying first 50 chars

# Clean up the dummy directory and files
import shutil
shutil.rmtree(dummy_dir)
print(f"\nCleaned up dummy directory '{dummy_dir}'.")
```
