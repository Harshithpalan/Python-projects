### Visualize Daily Water Usage Trends

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Generate sample daily water usage data
dates = pd.date_range(start='2023-01-01', periods=100, freq='D')
np.random.seed(42)
water_usage = np.random.normal(loc=150, scale=30, size=100)
# Add some seasonality or trend if desired (optional)
water_usage = water_usage + 10 * np.sin(np.linspace(0, 3 * np.pi, 100))
water_usage[water_usage < 0] = 0 # Ensure no negative usage

df_water = pd.DataFrame({'date': dates, 'daily_usage': water_usage})
display(df_water.head())
```

```python
# Plot daily water usage trends
fig, ax = plt.subplots(figsize=(14, 7))
sns.lineplot(x='date', y='daily_usage', data=df_water, ax=ax)
ax.set_title('Daily Water Usage Trends', fontsize=16)
ax.set_xlabel('Date', fontsize=12)
ax.set_ylabel('Water Usage (Liters)', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.6)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
