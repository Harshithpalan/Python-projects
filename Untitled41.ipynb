## Define Route Optimization Problem

### Subtask:
Formulate the route optimization problem by clearly defining the objective function and constraints. This formulation will serve as the foundation for developing the route optimization algorithm.


```markdown
### Route Optimization Problem Formulation

**Objective Functions:**

The primary goals of the route optimization problem are to minimize:

1.  **Total Distance Traveled:** Reduce the overall distance covered by all collection vehicles to complete their routes.
2.  **Total Fuel Consumption:** Minimize the amount of fuel used, which is directly related to distance and vehicle load.
3.  **Total Collection Time:** Optimize routes to reduce the time spent on collections, including travel and service time at each bin.
4.  **Operational Costs:** Indirectly, by minimizing distance, fuel, and time, the model aims to reduce overall operational costs associated with waste collection.

**Constraints:**

Several real-world limitations and rules must be adhered to during route optimization:

1.  **Vehicle Capacity:** Each vehicle has a maximum weight or volume capacity that cannot be exceeded by the collected waste.
2.  **Bin Locations:** All specified garbage bins requiring collection must be visited.
3.  **Predicted Fill Levels:** The predicted fill levels of each bin must be considered to prioritize collections and ensure bins are emptied before they overflow.
4.  **Maximum Route Duration/Driver Working Hours:** Each route must be completed within a practical time limit, respecting driver working hour regulations.
5.  **Time Windows (Optional but common):** Some bins or areas might have specific time windows during which collection is permitted.
6.  **Depot/Starting Point:** All routes must start and end at a designated depot.
7.  **Vehicle Availability:** The number of available collection vehicles is limited.
```

## Develop Route Optimization Algorithm

### Subtask:
Implement a route optimization algorithm (e.g., a variant of the Vehicle Routing Problem - VRP, or heuristics like genetic algorithms, simulated annealing, or ant colony optimization) to generate efficient collection routes. The algorithm will take into account the predicted fill levels, vehicle constraints, and the defined objective to suggest the most optimal paths.


**Reasoning**:
Aggregate the DataFrame by `bin_id` to get a single entry for each bin, representing its unique location, capacity, and a representative `fill_level_liters` (using the mean for simplicity). Then, define the depot location and vehicle capacities as placeholder values for the route optimization problem. Finally, extract the bin locations and their predicted fill levels to be used as 'customers' with 'demands' for the VRP.



````python
import numpy as np

# Aggregate the DataFrame by bin_id to get unique bin locations and a representative fill level.
# For simplicity with the dummy data (and since 'collection_date' was dropped),
# we'll use the mean fill_level_liters for each bin_id as the 'predicted' level.
# We also need to ensure that other bin-specific details like location and capacity are unique.
# Assuming location and capacity are constant per bin_id in the dummy data.
aggregated_bins = df.groupby('bin_id').agg({
    'location_lat': 'first',
    'location_lon': 'first',
    'capacity_liters': 'first',
    'fill_level_liters': 'mean' # Using mean as a representative 'predicted' fill level
}).reset_index()

print("Aggregated Bin Data (with predicted fill levels):")
print(aggregated_bins)

# 2. Define the depot location and a set of available vehicles with their capacities.
# Placeholder values for now.
depot_location = {'lat': 34.0520, 'lon': -118.2435} # A central point near the bins

# Define vehicles with their capacities in liters
vehicles = [
    {'id': 1, 'capacity_liters': 500},
    {'id': 2, 'capacity_liters': 700}
]
print(f"\nDepot Location: {depot_location}")
print(f"Available Vehicles: {vehicles}")

# 3. Extract bin locations and their corresponding predicted fill levels
# These will represent the 'customer' locations with 'demands' for the VRP.
bin_locations = aggregated_bins[['bin_id', 'location_lat', 'location_lon']].to_dict(orient='records')
bin_demands = aggregated_bins.set_index('bin_id')['fill_level_liters'].to_dict()

print("\nBin Locations (for VRP nodes):")
for bin_loc in bin_locations:
    print(bin_loc)
print("\nBin Demands (predicted fill levels in liters):")
print(bin_demands)
````

**Reasoning**:
Generate the distance matrix required for the VRP solver. This involves calculating the Haversine distance between all pairs of locations (depot and bins).



````python
from math import radians, sin, cos, sqrt, atan2

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Radius of Earth in kilometers

    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = R * c
    return distance

# Create a list of all locations, starting with the depot
v_locations = [depot_location]
v_locations.extend(bin_locations)

# Create a distance matrix (distance in km)
distance_matrix = np.zeros((len(v_locations), len(v_locations)))

for i in range(len(v_locations)):
    for j in range(len(v_locations)):
        if i == j:
            distance_matrix[i][j] = 0
        else:
            lat1 = v_locations[i]['location_lat'] if 'location_lat' in v_locations[i] else v_locations[i]['lat']
            lon1 = v_locations[i]['location_lon'] if 'location_lon' in v_locations[i] else v_locations[i]['lon']
            lat2 = v_locations[j]['location_lat'] if 'location_lat' in v_locations[j] else v_locations[j]['lat']
            lon2 = v_locations[j]['location_lon'] if 'location_lon' in v_locations[j] else v_locations[j]['lon']
            distance_matrix[i][j] = haversine(lat1, lon1, lat2, lon2)

print("Distance Matrix (km):")
print(distance_matrix)
````

**Reasoning**:
Set up the data model required by the OR-Tools VRP solver.



````python
data = {}
data['distance_matrix'] = (distance_matrix * 1000).astype(int) # Convert to meters and then to int for OR-Tools
# The demands correspond to the predicted fill levels for each bin.
# The depot (index 0) has a demand of 0.
data['demands'] = [0] + [bin_demands[b['bin_id']] for b in bin_locations]
data['vehicle_capacities'] = [v['capacity_liters'] for v in vehicles]
data['num_vehicles'] = len(vehicles)
data['depot'] = 0 # The depot is the first element in the distance matrix

print("VRP Data Model:")
print(f"  Distance Matrix (m, int):\n{data['distance_matrix']}")
print(f"  Demands (liters): {data['demands']}")
print(f"  Vehicle Capacities (liters): {data['vehicle_capacities']}")
print(f"  Number of Vehicles: {data['num_vehicles']}")
print(f"  Depot Index: {data['depot']}")
````

**Reasoning**:
The previous step successfully solved the VRP and output the objective value. Now, to understand the solution, it's essential to display the actual routes generated for each vehicle. This involves iterating through the solution to reconstruct the paths, including the sequence of bins visited, and the total distance and load for each route.



````python
def print_solution(data, manager, routing, solution):
    """Prints solution on console."""
    print(f"Objective: {solution.ObjectiveValue()} meters")
    total_distance = 0
    total_load = 0
    for vehicle_id in range(data['num_vehicles']):
        index = routing.Start(vehicle_id)
        route_distance = 0
        route_load = 0
        route_nodes = []
        while not routing.IsEnd(index):
            node_index = manager.IndexToNode(index)
            route_load += data['demands'][node_index]
            route_nodes.append(node_index)
            previous_index = index
            index = solution.Value(routing.NextVar(index))
            route_distance += routing.Get ;ArcCostForVehicle(previous_index, index, vehicle_id)
        route_nodes.append(manager.IndexToNode(index))
        print(f"Route for vehicle {vehicle_id + 1}:")
        print(f"  Route: {route_nodes}")
        print(f"  Distance of the route: {route_distance} meters")
        print(f"  Load of the route: {route_load} liters")
        total_distance += route_distance
        total_load += route_load
    print(f"Total distance of all routes: {total_distance} meters")
    print(f"Total load of all routes: {total_load} liters")

if solution:
    print_solution(data, manager, routing, solution)
else:
    print("No solution found.")
````

````python
import folium
import matplotlib.pyplot as plt # Keeping this import as it might be useful for other plots

def plot_routes_on_map(data, manager, routing, solution, bin_locations, depot_location, file_name='optimized_routes.html'):
    """Plots the optimized routes on an interactive Folium map."""

    # Get coordinates for all locations (depot + bins)
    coords = []
    # Depot coordinates
    coords.append((depot_location['lat'], depot_location['lon']))
    # Bin coordinates
    for bin_loc in bin_locations:
        coords.append((bin_loc['location_lat'], bin_loc['location_lon']))

    # Create a base map centered around the depot
    m = folium.Map(location=[depot_location['lat'], depot_location['lon']], zoom_start=13)

    # Add depot marker
    folium.Marker(
        location=[depot_location['lat'], depot_location['lon']],
        popup='Depot',
        icon=folium.Icon(color='red', icon='warehouse')
    ).add_to(m)

    # Add bin markers
    for i, bin_loc in enumerate(bin_locations):
        folium.Marker(
            location=[bin_loc['location_lat'], bin_loc['location_lon']],
            popup=f"Bin ID: {bin_loc['bin_id']}\nFill Level: {data['demands'][i+1]:.1f} L", # i+1 because data['demands'][0] is depot
            icon=folium.Icon(color='blue', icon='trash')
        ).add_to(m)

    colors = ['green', 'purple', 'orange', 'darkgreen', 'cadetblue', 'pink'] # More colors if needed

    if solution:
        for vehicle_id in range(data['num_vehicles']):
            index = routing.Start(vehicle_id)
            route_coords = []
            while not routing.IsEnd(index):
                node_index = manager.IndexToNode(index)
                route_coords.append(coords[node_index])
                index = solution.Value(routing.NextVar(index))
            # Add the last node (depot or bin)
            node_index = manager.IndexToNode(index)
            route_coords.append(coords[node_index])

            # Draw the route on the map
            if len(route_coords) > 1: # Only draw if there's an actual route
                folium.PolyLine(route_coords, color=colors[vehicle_id % len(colors)], weight=2.5, opacity=1).add_to(m)
                # Add start/end markers for the route (optional)
                folium.CircleMarker(location=route_coords[0], radius=5, color=colors[vehicle_id % len(colors)], fill=True, fill_color=colors[vehicle_id % len(colors)], popup=f'Vehicle {vehicle_id + 1} Start').add_to(m)
                folium.CircleMarker(location=route_coords[-1], radius=5, color=colors[vehicle_id % len(colors)], fill=True, fill_color=colors[vehicle_id % len(colors)], popup=f'Vehicle {vehicle_id + 1} End').add_to(m)


    # Save the map to an HTML file and display it
    m.save(file_name)
    print(f"Map saved to {file_name}")

print("plot_routes_on_map function defined.")
````

**Reasoning**:
Now that the `plot_routes_on_map` function is defined, I will call it with the generated VRP solution and location data to visualize the optimized routes on an interactive map. This will allow for a visual evaluation of the routes.



````python
plot_routes_on_map(data, manager, routing, solution, bin_locations, depot_location)
print("Routes visualized on map and saved to optimized_routes.html")
````
