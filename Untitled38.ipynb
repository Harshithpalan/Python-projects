## Smart Complaint Management System - NLP Complaint Classification

This section outlines a basic Python code structure for classifying complaints using Natural Language Processing (NLP). It covers data loading, preprocessing, model training, and prediction.

```python
# 1. Import necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download NLTK data (if not already downloaded).
# These downloads are required for `stopwords` and `WordNetLemmatizer`.
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
```

### Data Loading (Placeholder)

In a real-world scenario, you would load your complaint data from a CSV, database, or other source. For this example, we'll create a dummy DataFrame.

```python
# 2. Dummy Data Loading
data = {
    'complaint_text': [
        'Pothole on Main Street near the park entrance. It is quite large.',
        'Garbage overflowing in the bin on Elm Avenue.',
        'Water pipe leaking heavily on Oak Street since yesterday.',
        'Streetlight broken on Pine Lane, completely dark at night.',
        'Another pothole, this time on Maple Drive.',
        'Trash not collected from my house on Willow Road.',
        'Small water leak near the fire hydrant on Birch Street.',
        'Lights are out on Cedar Avenue. Safety concern.',
        'Very deep pothole on the corner of Market and 1st.',
        'Recycling bins are full and spilling over on Church Street.'
    ],
    'complaint_type': [
        'Pothole', 'Garbage', 'Water Leak', 'Streetlight', 'Pothole',
        'Garbage', 'Water Leak', 'Streetlight', 'Pothole', 'Garbage'
    ]
}
df = pd.DataFrame(data)

print("Original DataFrame head:")
display(df.head())
```

### Text Preprocessing

This involves cleaning the text data to prepare it for the NLP model. Common steps include lowercasing, removing punctuation, numbers, stopwords, and lemmatization.

```python
# 3. Text Preprocessing Function
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    text = text.lower() # Lowercase
    text = re.sub(r'[^a-z\s]', '', text) # Remove punctuation and numbers
    words = text.split() # Tokenize
    words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words] # Remove stopwords and lemmatize
    return ' '.join(words)

df['processed_complaint_text'] = df['complaint_text'].apply(preprocess_text)

print("\nDataFrame after preprocessing:")
display(df[['complaint_text', 'processed_complaint_text']].head())
```

### Model Training

Here we will split the data into training and testing sets, vectorise the text using TF-IDF, and train a classification model (e.g., Logistic Regression).

```python
# 4. Model Training

X = df['processed_complaint_text']
y = df['complaint_type']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42, stratify=y)

# Create a pipeline with TF-IDF vectorizer and a Logistic Regression classifier
model_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('classifier', LogisticRegression(random_state=42, solver='liblinear'))
])

# Train the model
model_pipeline.fit(X_train, y_train)

# Evaluate the model
y_pred = model_pipeline.predict(X_test)
print("\nClassification Report on Test Data:")
print(classification_report(y_test, y_pred))
```

### Prediction Function

A function to take new, unseen complaint text and classify it using the trained model.

```python
# 5. Prediction Function
def predict_complaint_type(new_complaint_text):
    processed_text = preprocess_text(new_complaint_text)
    prediction = model_pipeline.predict([processed_text])
    return prediction[0]

# Example Usage:
new_complaint1 = "There's a huge hole in the road near the school."
predicted_type1 = predict_complaint_type(new_complaint1)
print(f"\nNew complaint: '{new_complaint1}' -> Predicted type: '{predicted_type1}'")

new_complaint2 = "The street lamp outside my house is not working."
predicted_type2 = predict_complaint_type(new_complaint2)
print(f"New complaint: '{new_complaint2}' -> Predicted type: '{predicted_type2}'")

new_complaint3 = "Water is gushing out of a pipe on the sidewalk."
predicted_type3 = predict_complaint_type(new_complaint3)
print(f"New complaint: '{new_complaint3}' -> Predicted type: '{predicted_type3}'")
```
